### Z-Score Normalization

The Z-score normalization formula is:

$$
z = \frac{x-\mu}{\sigma}
$$

Where:
- $x$ = Original data value
- $\mu$ = Mean of the dataset
- $\sigma$ = Standard deviation of the dataset
- $z$ = Normalized Z-score

```python
## Standardizes the features X and target variable y using Z-score normalization (also called Standardization). 
X = (X - np.mean(X, axis=0)) / np.std(X, axis=0)
Y = (Y - np.mean(Y, axis=0)) / np.std(Y, axis=0)
```

[![Single-Input-Linear-Neuron-Diagram.png](https://i.postimg.cc/PxYWMz4F/Single-Input-Linear-Neuron-Diagram.png)](https://postimg.cc/vg8xYnCW)

```python
X = np.hstack((np.ones((X.shape[0], 1)), X))
print('Design matrix shape:', X.shape)
```

```python
#weight or parameter initialization
theta = np.zeros(X.shape[1])
iterations = 1000
learningrate = 0.01
mseh = []
m = X.shape[0]

for _ in range(iterations):
    y_pred = X @ theta
    error = y_pred - Y
    mse = np.mean(error ** 2)
    mseh.append(mse)
    gradient = (2 / m) * (X.T @ error)
    theta -= learningrate * gradient

print('Learned parameters [bias, height coefficient]:', theta)
print('Final training MSE:', mseh[-1])
```

```python
y_pred = X @ theta
manual_mse = mean_squared_error(Y, y_pred)
manual_r2 = r2_score(Y, y_pred)
print('Manual gradient-descent MSE:', manual_mse)
print('Manual gradient-descent R-squared:', manual_r2)
```

```python
# Plot the MSE over iterations
plt.plot(range(iterations), mseh)
plt.xlabel('Iterations')
plt.ylabel('Mean Squared Error (MSE)')
plt.title('MSE vs Iterations')
plt.show()
```

## You are required to fit a linear function on Food_delivery_times dataset
Description about the dataset: This dataset is designed for predicting food delivery times based on various influencing factors such as distance, weather, traffic conditions, and Preperation time. 
1. Feature selection: Consider Food_delivery_time is effected by four features (attributes) given in the dataset, which are Distance_km, Preparation_Time_min, Traffic, Weather
2. Since Traffic, Weather are having categorical values, you need to convert them into numerical using the categorical encoding methods.
3. Now, fit a linear regression model considering Distance_km, Preparation_Time_min, Traffic, and Weather as independant variables and Delivery_Time_min as dependant varibal. 


Note: In lab record, you can write the statements relevant to linear regression.

```python
food_path = 'Food_Delivery_Times.csv'
df_food = pd.read_csv(food_path)
print('Dataset shape:', df_food.shape)
df_food.head()
```

```python
# Keep only the four features required by the assignment
food = df_food[['Distance_km', 'Preparation_Time_min', 'Traffic_Level', 'Weather', 'Delivery_Time_min']].copy()

# One-hot encode Traffic_Level and Weather; drop_first avoids redundant dummy columns.
food_encoded = pd.get_dummies(
    food,
    columns=['Traffic_Level', 'Weather'],
    drop_first=True,
    dtype=float
)

X_food = food_encoded.drop(columns='Delivery_Time_min')
y_food = food_encoded['Delivery_Time_min']
print('Encoded feature columns:', list(X_food.columns))
print('X shape:', X_food.shape, '| y shape:', y_food.shape)
X_food.head()
```

```python
X_food_train, X_food_test, y_food_train, y_food_test = train_test_split(
    X_food, y_food, test_size=0.20, random_state=67
)

food_model = LinearRegression()
food_model.fit(X_food_train, y_food_train)
y_food_pred = food_model.predict(X_food_test)

food_mse = mean_squared_error(y_food_test, y_food_pred)
food_rmse = np.sqrt(food_mse)
food_mae = mean_absolute_error(y_food_test, y_food_pred)
food_r2 = r2_score(y_food_test, y_food_pred)

print(f'Mean Squared Error (MSE): {food_mse:.4f}')
print(f'Root Mean Squared Error (RMSE): {food_rmse:.4f}')
print(f'Mean Absolute Error (MAE): {food_mae:.4f}')
print(f'R-squared (R²): {food_r2:.4f}')

coefficients = pd.Series(food_model.coef_, index=X_food.columns)
print('\nIntercept:', round(food_model.intercept_, 4))
print('Coefficients:')
print(coefficients.round(4))
```

```python
# Plot actual versus predicted delivery times
plt.figure(figsize=(7, 5))
plt.scatter(y_food_test, y_food_pred, alpha=0.65, edgecolor='k', linewidth=0.3)
line_min = min(y_food_test.min(), y_food_pred.min())
line_max = max(y_food_test.max(), y_food_pred.max())
plt.plot([line_min, line_max], [line_min, line_max], 'r--', label='Perfect prediction')
plt.xlabel('Actual Delivery Time (min)')
plt.ylabel('Predicted Delivery Time (min)')
plt.title('Food Delivery: Actual vs Predicted Time')
plt.legend()
plt.show()
```
