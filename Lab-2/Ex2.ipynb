## Step 6  Option 1: Identifying Outliers Using Z-Scores
The value of 3 in the context of Z-scores is often used as a threshold to identify outliers in a dataset. A Z-score represents how many standard deviations a data point is away from the mean of the dataset. Specifically:

A Z-score of 0 means the data point is exactly at the mean.
A Z-score of 1 means the data point is one standard deviation above the mean, and so on.
A Z-score of 3 corresponds to a data point being 3 standard deviations away from the mean. For a normal distribution, about 99.7% of the data points fall within 3 standard deviations of the mean (according to the 68-95-99.7 rule, which describes the spread of data in a normal distribution). Therefore, points with Z-scores greater than 3 or less than -3 are considered unusually far from the mean and are often flagged as outliers.

This threshold (Z > 3 or Z < -3) is commonly used in many statistical applications because it captures the extreme values that are rare in a normal distribution, which are typically considered to be outliers. However, the choice of threshold can vary depending on the specific application and the nature of the data.

```python
import pandas as pd
import numpy as np
from scipy import stats
from sklearn.preprocessing import LabelEncoder, StandardScaler, MinMaxScaler

# ── STEP 1: Load & Explore ────────────────────────────────────────────────────
df = pd.read_csv('Automobile.csv')
print("=== FIRST 10 ROWS ==="); display(df.head(10))
print("\nShape:", df.shape)
print("\nColumns:", df.columns.tolist())
print("\nData Types:\n", df.dtypes)
print("\nDescriptive Statistics:"); display(df.describe(include='all'))

numerical_cols = df.select_dtypes(include=['int64','float64']).columns.tolist()
categorical_cols = df.select_dtypes(include=['str','object']).columns.tolist()
print("\nNumerical Columns:", numerical_cols)
print("Categorical Columns:", categorical_cols)
for col in categorical_cols:
    print(f"\nUnique values in '{col}':", df[col].unique())

# ── STEP 2: Data Cleaning ─────────────────────────────────────────────────────
print("\n=== MISSING VALUES ===")
missing_df = pd.DataFrame({'Count': df.isnull().sum(), 'Percentage': (df.isnull().sum()/len(df))*100})
display(missing_df[missing_df['Count'] > 0])

# Fix horsepower FIRST (has '?' strings)
df['horsepower'] = pd.to_numeric(df['horsepower'].replace('?', np.nan), errors='coerce')

# Refresh numerical columns after fixing horsepower
numerical_cols = df.select_dtypes(include=['int64','float64']).columns.tolist()
categorical_cols = df.select_dtypes(include=['str','object']).columns.tolist()

# Fill missing values (pandas 2.x compatible, no inplace)
for col in numerical_cols:
    if df[col].isnull().sum() > 0:
        df[col] = df[col].fillna(df[col].median())
for col in categorical_cols:
    if df[col].isnull().sum() > 0:
        df[col] = df[col].fillna(df[col].mode()[0])

print("\nMissing values after handling:", df.isnull().sum().sum())

# Duplicates
print("\nDuplicate rows:", df.duplicated().sum())
df = df.drop_duplicates()
print("After removing duplicates — Shape:", df.shape)

# ── STEP 3: Data Transformation ───────────────────────────────────────────────
# origin is already 'usa'/'europe'/'japan' as strings — just verify
print("\nOrigin value counts:\n", df['origin'].value_counts())

# New columns
df['weight_kg']    = df['weight'] * 0.453592
df['mpg_category'] = pd.cut(df['mpg'], bins=[0, 20, 30, np.inf], labels=['Low','Medium','High'])
df['vehicle_age']  = 2026 - (1900 + df['model_year'])   # model_year is 70,71,...82 → 1970..1982
print("\nNew columns sample:"); display(df[['mpg','weight','weight_kg','mpg_category','model_year','vehicle_age']].head())

# Rename
df = df.rename(columns={'mpg':'miles_per_gallon','horsepower':'hp','weight':'weight_lbs','model_year':'year'})
print("\nRenamed Columns:", df.columns.tolist())

# Filters
print("\n── Vehicles with miles_per_gallon > 30 ──");  display(df[df['miles_per_gallon'] > 30])
print("\n── Vehicles with hp > 150 ──");               display(df[df['hp'] > 150])
print("\n── Vehicles with cylinders >= 6 ──");         display(df[df['cylinders'] >= 6])
print("\n── Vehicles manufactured after 1980 ──");     display(df[df['year'] > 80])
print("\n── Vehicles from USA ──");                    display(df[df['origin'] == 'usa'])

# ── STEP 4: Encoding ──────────────────────────────────────────────────────────
le = LabelEncoder()
df['origin_encoded'] = le.fit_transform(df['origin'])
print("\n=== LABEL ENCODING ===")
display(df[['origin','origin_encoded']].drop_duplicates().sort_values('origin_encoded'))
print("Mapping:", {cat: label for label, cat in enumerate(le.classes_)})

ohe_df = pd.get_dummies(df['origin'], prefix='origin')
print("\n=== ONE-HOT ENCODING ==="); display(ohe_df.head())
print("""
Which is better for 'origin'?
→ One-Hot Encoding is preferred.
  'origin' is NOMINAL (no order among usa/europe/japan).
  Label Encoding falsely implies europe(0) < japan(1) < usa(2).
  One-Hot avoids this by creating separate binary columns.
""")

# ── STEP 5: Outlier Detection ─────────────────────────────────────────────────
num_cols = df.select_dtypes(include=['int64','float64']).columns.tolist()
z_scores = pd.DataFrame(np.abs(stats.zscore(df[num_cols].dropna())), columns=num_cols)
outlier_mask = (z_scores > 3).any(axis=1)
print("\n=== OUTLIERS (|Z| > 3) ===")
print("Outlier count per column:\n", (z_scores > 3).sum())
print(f"\nTotal outlier rows: {outlier_mask.sum()}")
display(df.iloc[z_scores.index[outlier_mask]])
print("Decision: RETAIN — extreme values represent real vehicle specs, not data errors.")

# ── STEP 6: Standardization ───────────────────────────────────────────────────
num_cols_clean = [c for c in num_cols if df[c].isnull().sum() == 0]
df_std = pd.DataFrame(StandardScaler().fit_transform(df[num_cols_clean]), columns=num_cols_clean)
print("\n=== STANDARDIZED (first 5 rows) ==="); display(df_std.head())
print("Mean & Std:"); display(df_std.describe().loc[['mean','std']].round(4))

# ── STEP 7: Normalization ─────────────────────────────────────────────────────
df_norm = pd.DataFrame(MinMaxScaler().fit_transform(df[num_cols_clean]), columns=num_cols_clean)
print("\n=== NORMALIZED (first 5 rows) ==="); display(df_norm.head())
print("Min & Max:"); display(df_norm.describe().loc[['min','max']].round(4))
print("""
Standardization (StandardScaler) → mean=0, std=1
  Use for: Linear/Logistic Regression, PCA, SVM, normal distributions.

Normalization (MinMaxScaler) → range [0,1]
  Use for: KNN, Neural Networks, when feature magnitudes vary widely.
""")

# ── STEP 8: X and Y ───────────────────────────────────────────────────────────
feature_cols = ['cylinders','displacement','hp','weight_lbs','acceleration','year','weight_kg','vehicle_age']
X = df[feature_cols]
Y = df['miles_per_gallon']
print("\n=== X and Y ===")
print("X shape:", X.shape, "| Y shape:", Y.shape)
display(X.head())

pd.concat([X.reset_index(drop=True), Y.reset_index(drop=True)], axis=1).to_csv('automobile_X_Y.csv', index=False)
print("\nLoaded automobile_X_Y.csv:"); display(pd.read_csv('automobile_X_Y.csv').head())

# ── STEP 9: Save Final Dataset ────────────────────────────────────────────────
final_df = pd.concat([X.reset_index(drop=True), Y.reset_index(drop=True)], axis=1)
print("\n=== FINAL DATASET ==="); display(final_df.head())
print("Shape:", final_df.shape)
print("Missing values:", final_df.isnull().sum().sum())
final_df.to_csv('automobile_preprocessed.csv', index=False)
print("Saved as automobile_preprocessed.csv ✓")
```
