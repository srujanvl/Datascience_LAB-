# Descriptive Statistics on a Synthetic Employee Dataset

**Aim:** To create a synthetic dataset and perform a complete descriptive analysis on it.

**Everything is inside this one notebook.** The dataset is generated in Part 1, so there is
no file to download. Running the notebook from top to bottom reproduces every number exactly.

### What we will calculate

| Part | Topic | Functions used |
|---|---|---|
| 3 | Mean, Median, Mode | `.mean()`, `.median()`, `.mode()` |
| 4 | Range, Variance, Std Dev, IQR | `.max()-.min()`, `.var()`, `.std()`, `.quantile()` |
| 5 | Skewness, Kurtosis | `.skew()`, `.kurt()` |
| 6 | Visualization | Seaborn |
| 7 | Outlier detection | IQR rule, Z-score rule |
| 8 | Covariance | `.cov()` |
| 9 | Correlation | `.corr()` (Pearson and Spearman) |

### Why a synthetic dataset?

Because we **build** the relationships ourselves, we know the right answer in advance.
When the analysis later reports that experience and salary are strongly related, we can confirm
that this is genuinely what we put into the data, not an accident. That makes it a good way
to learn what each statistic is actually detecting.

# Part 1: Creating the synthetic dataset

We simulate an **HR database of 400 employees** at a software company.

Each column is designed on purpose to demonstrate one idea:

| Column | Type | Designed to show |
|---|---|---|
| `department`, `gender`, `city`, `education` | Text | **Mode** (the only average that works on text) |
| `age`, `work_hours_per_week` | Normal | Mean ≈ median ≈ mode, low skewness |
| `experience_years`, `commute_km` | Right-skewed | Mean **greater** than median |
| `performance_score` | Left-skewed | Mean **less** than median |
| `satisfaction_score` | Near-symmetric | A case where the two skew tests disagree |
| `projects_completed` | Count (Poisson) | Discrete whole-number data |
| `annual_salary_inr` | Strong right skew + outliers | Outlier detection, robust vs non-robust statistics |

We also build in specific relationships so the correlation section has something real to find:

- `age` and `experience_years` should be **very strongly** related (you cannot have 20 years of experience at 25).
- `experience_years` should **drive salary** strongly.
- `work_hours_per_week` should **reduce** satisfaction — a **negative** relationship.
- `commute_km` should also reduce satisfaction, but more weakly.
- `commute_km` and salary should have **no** relationship at all.

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats

sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (9, 5)
plt.rcParams["axes.titlesize"] = 12
pd.set_option("display.width", 120)
pd.set_option("display.max_columns", 30)

rng = np.random.default_rng(2024)      # fixed seed -> same data every time
N = 400
print("Libraries loaded. Generating", N, "employee records...")
```

```python
# ---------- 1. Text columns (for the MODE) ----------
department = rng.choice(["IT", "Sales", "Operations", "Finance", "HR"],
                        N, p=[.32, .24, .20, .14, .10])
gender = rng.choice(["Male", "Female"], N, p=[.58, .42])
city = rng.choice(["Bengaluru", "Hyderabad", "Pune", "Chennai", "Mysuru"],
                  N, p=[.30, .22, .20, .16, .12])
education = rng.choice(["Diploma", "Bachelor", "Master", "PhD"],
                       N, p=[.14, .50, .30, .06])

# ---------- 2. Roughly normal columns ----------
age = rng.normal(34, 8, N).clip(22, 60)
work_hours = rng.normal(44, 6, N).clip(30, 72)
training_hours = rng.normal(30, 11, N).clip(0, 90)

# ---------- 3. Experience depends on age -> strong positive correlation ----------
experience = ((age - 22) * rng.uniform(0.45, 0.95, N)).clip(0, 35)

# ---------- 4. Left-skewed columns (most people score well, a few score poorly) ----------
performance = (9.0 - rng.gamma(2.6, 1.05, N) + 0.030 * training_hours).clip(1, 10)

# ---------- 5. Count and heavy-tailed columns ----------
projects = rng.poisson(np.clip(experience / 3.2 + 1.5, 0.4, None)).clip(0, 25)
commute = rng.exponential(9, N).clip(0.5, 60)

# ---------- 6. Satisfaction FALLS with long hours and long commute (negative link) ----------
satisfaction = (7.2
                - 0.115 * (work_hours - 40)      # negative coefficient
                - 0.048 * commute                # negative coefficient
                + 0.14 * performance
                + rng.normal(0, 1.05, N)).clip(1, 10)

# ---------- 7. Salary driven mainly by experience ----------
edu_points = pd.Series(education).map(
    {"Diploma": 0, "Bachelor": 1, "Master": 2, "PhD": 3}).values

salary = (280000
          + 72000 * experience          # experience is the dominant driver
          + 95000 * edu_points
          + 32000 * performance
          + 1500 * projects
          + rng.normal(0, 120000, N))
salary = (salary * rng.lognormal(0, 0.16, N)).clip(220000, None)

print("All columns generated.")
```

```python
df = pd.DataFrame({
    "emp_id": [f"EMP{1000+i}" for i in range(N)],
    "department": department,
    "gender": gender,
    "city": city,
    "education": education,
    "age": np.round(age, 0).astype(int),
    "experience_years": np.round(experience, 1),
    "training_hours": np.round(training_hours, 0).astype(int),
    "performance_score": np.round(performance, 1),
    "work_hours_per_week": np.round(work_hours, 1),
    "projects_completed": projects,
    "commute_km": np.round(commute, 1),
    "satisfaction_score": np.round(satisfaction, 1),
    "annual_salary_inr": np.round(salary, -2),
})

# ---------- 8. Add 5 senior executives as deliberate OUTLIERS ----------
# These are real employees, not errors. Part 7 will detect them.
exec_rows = rng.choice(N, 5, replace=False)
df.loc[exec_rows, "annual_salary_inr"] = [6850000, 5920000, 8100000, 5400000, 7250000]
df.loc[exec_rows, "experience_years"] = [28.5, 26.0, 31.0, 25.5, 29.0]

df.to_csv("employee_data.csv", index=False)      # saved for reference
print("Dataset created:", df.shape[0], "rows x", df.shape[1], "columns")
print("Saved as employee_data.csv")
df.head()
```

### How to read `describe()`

- **count** — how many non-missing values. All 400 here, so nothing is missing.
- **mean** — the average.
- **std** — the standard deviation.
- **min / max** — the smallest and largest values.
- **25% / 50% / 75%** — the first quartile (Q1), the median, and the third quartile (Q3).

Already one thing stands out: for `annual_salary_inr` the **max is far above the 75% value**.
That is the first sign of outliers, which we will confirm in Part 7.

`describe()` does **not** give us mode, variance, range, IQR, skewness or kurtosis.
We calculate all of those in the next parts.

# Part 3: Measures of central tendency

These answer the question **"what is a typical value?"**

## 3.1 Mean

$$\bar{x} = \frac{x_1 + x_2 + \dots + x_n}{n} = \frac{\sum x_i}{n}$$

Add every value and divide by how many there are. The mean uses **all** the data, which is its
strength and also its weakness — a single extreme value affects it.

## 3.2 Median

Sort the values and take the middle one. If there is an even number of values, average the
middle two. The median is the **50th percentile**: half the data lies below it.

Because it only cares about *position*, not size, one huge value cannot move it far.
That makes the median **robust**.

## 3.3 Mode

The value that occurs most often. It is the **only** measure of centre that works on text data
such as `department` or `city`.

```python
# Start with one column so the idea is clear
col = df["annual_salary_inr"]

print("Column: annual_salary_inr")
print("-" * 45)
print(f"Mean   : {col.mean():>12,.2f}")
print(f"Median : {col.median():>12,.2f}")
print(f"Mode   : {col.mode()[0]:>12,.2f}")
```

```python
# All numeric columns together
central = pd.DataFrame({
    "Mean": df[num_cols].mean(),
    "Median": df[num_cols].median(),
    "Mode": df[num_cols].mode().iloc[0],
})
central["Mean - Median"] = central["Mean"] - central["Median"]
central.round(2)
```

### Interpreting the "Mean − Median" column

This single number tells us the direction of the skew:

- **Positive** → the mean has been pulled **up** by large values → **right-skewed**.
- **Negative** → the mean has been pulled **down** by small values → **left-skewed**.
- **Near zero** → the data is balanced.

Look at `annual_salary_inr`: the mean is far above the median. Those five executives are
dragging the average up, so **the mean salary is not a typical salary**.

Now look at `performance_score`: the mean is *below* the median. Most employees score well,
and a small number of low scorers pull the average down. That is left skew.

```python
# Proof that the median is robust and the mean is not
salary = df["annual_salary_inr"]
no_exec = salary.nsmallest(len(salary) - 5)     # drop the 5 highest salaries

print("                     With executives    Without executives    Change")
print("-" * 72)
print(f"Mean            {salary.mean():>15,.0f} {no_exec.mean():>20,.0f} "
      f"{(no_exec.mean()-salary.mean())/salary.mean()*100:>10.1f}%")
print(f"Median          {salary.median():>15,.0f} {no_exec.median():>20,.0f} "
      f"{(no_exec.median()-salary.median())/salary.median()*100:>10.1f}%")
```

Removing just **5 employees out of 400** moves the mean by several percent
but barely touches the median. This is the single most important practical lesson
in descriptive statistics.

```python
# Mode is the ONLY option for text columns
for c in cat_cols:
    counts = df[c].value_counts()
    print(f"{c:<12} mode = {df[c].mode()[0]:<12} ({counts.iloc[0]} employees, "
          f"{counts.iloc[0]/len(df)*100:.1f}%)")
```

```python
# Full frequency table for department
df["department"].value_counts().to_frame("count").assign(
    percent=lambda t: (t["count"] / len(df) * 100).round(1))
```

# Part 4: Measures of dispersion

Two departments can have the same average salary and still be completely different — one
where everybody earns about the same, another with a few very high and very low earners.
Dispersion measures that difference.

## 4.1 Range

$$\text{Range} = \text{Maximum} - \text{Minimum}$$

The simplest measure, but it uses only two values, so a single outlier decides it entirely.

## 4.2 Variance

$$s^{2} = \frac{\sum (x_i - \bar{x})^{2}}{n - 1}$$

For each value, measure how far it is from the mean, square it (so positives and negatives
do not cancel out), and average the results.

**Why divide by $n-1$ and not $n$?**
A sample's own mean always sits closer to its own points than the true population mean does,
so the squared distances come out slightly too small. Dividing by $n-1$ corrects this bias.
This is called **Bessel's correction**. `pandas` does it automatically.

## 4.3 Standard deviation

$$s = \sqrt{s^{2}}$$

The problem with variance is its **unit**: if salary is in rupees, variance is in rupees².
Taking the square root brings it back to rupees, so we can compare it directly against the mean.

## 4.4 Interquartile Range (IQR)

$$IQR = Q_3 - Q_1$$

The range of the **middle 50%** of the data. It ignores the top 25% and the bottom 25%
entirely, which makes it **robust** — the natural partner of the median.

```python
col = df["annual_salary_inr"]

Q1 = col.quantile(0.25)
Q3 = col.quantile(0.75)

print("Column: annual_salary_inr")
print("-" * 50)
print(f"Minimum            : {col.min():>14,.2f}")
print(f"Maximum            : {col.max():>14,.2f}")
print(f"Range              : {col.max() - col.min():>14,.2f}")
print()
print(f"Variance           : {col.var():>14,.2f}   (rupees squared)")
print(f"Standard deviation : {col.std():>14,.2f}   (rupees)")
print()
print(f"Q1 (25th pct)      : {Q1:>14,.2f}")
print(f"Q3 (75th pct)      : {Q3:>14,.2f}")
print(f"IQR (Q3 - Q1)      : {Q3 - Q1:>14,.2f}")
```

```python
# All columns at once
spread = pd.DataFrame({
    "Min": df[num_cols].min(),
    "Max": df[num_cols].max(),
    "Range": df[num_cols].max() - df[num_cols].min(),
    "Variance": df[num_cols].var(),
    "Std Dev": df[num_cols].std(),
    "Q1": df[num_cols].quantile(0.25),
    "Q3": df[num_cols].quantile(0.75),
})
spread["IQR"] = spread["Q3"] - spread["Q1"]

# Coefficient of variation = relative spread, unit-free
spread["CV %"] = (df[num_cols].std() / df[num_cols].mean() * 100)
spread.round(2)
```

### The Coefficient of Variation

We cannot say "salary varies more than age" by comparing standard deviations, because one is
in rupees and the other in years. The **coefficient of variation** solves this:

$$CV = \frac{s}{\bar{x}} \times 100\%$$

It expresses the spread as a percentage of the mean, so it is unit-free and comparable across
any two columns. A high CV means the values are widely scattered relative to their average.

### Reading a standard deviation

Take `work_hours_per_week`. If the mean is about 44 with a standard deviation of about 6,
then a typical employee works roughly 6 hours away from 44 — that is, most people are between
38 and 50 hours. Someone working 62 hours is three standard deviations out, which is rare.

```python
# The empirical rule: for roughly normal data,
# ~68% of values lie within 1 SD of the mean, ~95% within 2 SD, ~99.7% within 3 SD
check_col = "work_hours_per_week"
m, s = df[check_col].mean(), df[check_col].std()

print(f"Column: {check_col}   (mean = {m:.2f}, std = {s:.2f})")
print("-" * 62)
for k, theory in [(1, 68.0), (2, 95.0), (3, 99.7)]:
    pct = df[check_col].between(m - k*s, m + k*s).mean() * 100
    print(f"  Within +/- {k} SD  ({m-k*s:6.2f} to {m+k*s:6.2f}) : "
          f"observed {pct:6.2f}%   theory {theory}%")
```

# Part 5: Shape — skewness and kurtosis

Mean and standard deviation describe *where* the data sits and *how spread out* it is.
Skewness and kurtosis describe its **shape**.

## 5.1 Skewness

Skewness measures **asymmetry** — whether one tail is longer than the other.

$$G_1 = \frac{n}{(n-1)(n-2)} \sum \left(\frac{x_i - \bar{x}}{s}\right)^{3}$$

The deviations are cubed, so the **sign is preserved**: values far above the mean produce large
positive contributions, values far below produce large negative ones.

| Skewness | Shape | Relationship |
|---|---|---|
| Greater than 0 | Long tail on the **right** | mean > median |
| Equal to 0 | Symmetric | mean = median |
| Less than 0 | Long tail on the **left** | mean < median |

Rule of thumb: below 0.5 is fairly symmetric, 0.5 to 1 is moderate, above 1 is strong.

## 5.2 Kurtosis

Kurtosis measures the **tails** — how often extreme values occur.

$$G_2 = \frac{n(n+1)}{(n-1)(n-2)(n-3)} \sum \left(\frac{x_i-\bar{x}}{s}\right)^{4}
- \frac{3(n-1)^{2}}{(n-2)(n-3)}$$

The deviations are raised to the **fourth** power, so distant points dominate completely.
`pandas` reports **excess kurtosis**, meaning a perfect normal curve gives **0**.

| Kurtosis | Meaning |
|---|---|
| Greater than 0 | Heavy tails — outliers are more common than normal (**leptokurtic**) |
| Equal to 0 | Same as a normal curve (**mesokurtic**) |
| Less than 0 | Light tails, flatter — often two groups mixed (**platykurtic**) |

```python
shape = pd.DataFrame({
    "Mean": df[num_cols].mean(),
    "Median": df[num_cols].median(),
    "Skewness": df[num_cols].skew(),
    "Kurtosis": df[num_cols].kurt(),
})

def describe_skew(g):
    if g > 1:    return "Strong RIGHT skew"
    if g > 0.5:  return "Moderate right skew"
    if g < -1:   return "Strong LEFT skew"
    if g < -0.5: return "Moderate left skew"
    return "Approximately symmetric"

def describe_kurt(k):
    if k > 1:    return "Heavy tails"
    if k < -1:   return "Very flat"
    return "Close to normal"

shape["Shape"] = shape["Skewness"].apply(describe_skew)
shape["Tails"] = shape["Kurtosis"].apply(describe_kurt)
shape.round(3).sort_values("Skewness")
```

### What this table tells us about our data

- **`annual_salary_inr`** has by far the largest skewness and an extremely large kurtosis.
  Both are caused by the same five executives. A very high kurtosis is a strong hint that
  outliers are present, *before* we run any outlier test.
- **`commute_km`** is strongly right-skewed. Most employees live close by, a few live very far.
  This is typical of any "distance" or "waiting time" measurement.
- **`performance_score`** is left-skewed — the tail points to the **low** side, because most
  employees score well and only a few score badly.
- **`satisfaction_score`** sits very close to zero on both measures. It is essentially symmetric.
- **`work_hours_per_week`** and **`age`** have skewness near zero, exactly as designed.

Notice the agreement with Part 3: columns with a positive "Mean − Median" also have positive
skewness. They are two different ways of measuring the same asymmetry.

```python
# Do the two ways of measuring asymmetry agree?
agreement = pd.DataFrame({
    "Mean - Median": df[num_cols].mean() - df[num_cols].median(),
    "Skewness": df[num_cols].skew(),
})

# A column is only meaningfully skewed if BOTH measures are clearly away from zero
too_close = (agreement["Skewness"].abs() < 0.25)

agreement["Verdict"] = np.where(
    too_close, "Too close to call",
    np.where(np.sign(agreement["Mean - Median"]) == np.sign(agreement["Skewness"]),
             "Agree", "Disagree"))

agreement.round(3).sort_values("Skewness")
```

### When the two measures disagree

Look at `satisfaction_score`. Its skewness is slightly **negative**, but its "Mean − Median" is
slightly **positive**. Do the two measures contradict each other?

No — both numbers are essentially **zero**. The column is symmetric, and when a distribution is
symmetric the tiny leftover difference is just random noise. That is why the code above marks
any column with $|skew| < 0.25$ as **"Too close to call"** instead of forcing a direction.

**The lesson:** never read a direction into a skewness value that is already close to zero.
Only trust the sign when the magnitude is meaningful.

# Part 6: Visualization

> **Always plot the data before trusting the numbers.**
> **Anscombe's quartet** is four datasets with identical means, variances and correlations
> that look completely different when plotted.

## 6.1 Histograms — the shape of each column

```python
fig, axes = plt.subplots(3, 3, figsize=(15, 10))

for ax, c in zip(axes.flatten(), num_cols):
    sns.histplot(data=df, x=c, bins=25, kde=True, ax=ax, color="#02707E")
    ax.set_title(f"{c}\nskew = {df[c].skew():.2f}", fontsize=10)
    ax.set_xlabel("")

plt.suptitle("Distribution of every numeric column", fontsize=14, y=1.01)
plt.tight_layout()
plt.show()
```

The salary histogram is almost unreadable because the five executives stretch the axis.
That itself is useful information. Let us zoom in on the main body of the data.

## 6.2 Mean versus median on skewed columns

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 4.8))

for ax, c, cut in [(axes[0], "annual_salary_inr", 0.98),
                   (axes[1], "performance_score", 1.0)]:
    sns.histplot(data=df, x=c, bins=30, kde=True, ax=ax, color="#02707E")
    ax.axvline(df[c].mean(), color="red", ls="--", lw=2,
               label=f"Mean = {df[c].mean():,.2f}")
    ax.axvline(df[c].median(), color="green", lw=2,
               label=f"Median = {df[c].median():,.2f}")
    if cut < 1:
        ax.set_xlim(0, df[c].quantile(cut))
        ax.set_xlabel(f"{c}   (zoomed to the lower 98%)")
    ax.set_title(f"{c}   |   skew = {df[c].skew():.2f}")
    ax.legend()

plt.tight_layout()
plt.show()
```

**Left graph (right-skewed):** the red mean line sits to the **right** of the green median line.

**Right graph (left-skewed):** the mean sits to the **left** of the median.

This is the visual version of the "Mean − Median" column from Part 3.

## 6.3 Box plots — the five-number summary

```python
fig, axes = plt.subplots(1, 3, figsize=(14, 4.5))

for ax, c in zip(axes, ["annual_salary_inr", "commute_km", "work_hours_per_week"]):
    sns.boxplot(y=df[c], ax=ax, color="#00A896", width=0.4)
    ax.set_title(c)
    ax.set_ylabel("")

plt.suptitle("Box plots — the dots are candidate outliers", fontsize=13, y=1.03)
plt.tight_layout()
plt.show()
```

A box plot draws five numbers: minimum, Q1, median, Q3, maximum.
The **box height is the IQR**. The whiskers extend to 1.5 × IQR, and every point beyond them
is drawn individually as a **candidate outlier** — exactly what we calculate in Part 7.

## 6.4 Comparing groups

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 4.8))

order = df.groupby("department")["annual_salary_inr"].median().sort_values().index
sns.boxplot(data=df, x="department", y="annual_salary_inr", order=order, ax=axes[0])
axes[0].set_title("Salary by department")
axes[0].set_ylim(0, df["annual_salary_inr"].quantile(0.97))

sns.countplot(data=df, x="department", order=df["department"].value_counts().index, ax=axes[1])
axes[1].set_title("Number of employees per department")

plt.tight_layout()
plt.show()
```

```python
# Group statistics as a table
df.groupby("department")[["annual_salary_inr", "experience_years", "satisfaction_score"]] \
  .agg(["mean", "median", "std"]).round(1)
```

Notice that within each department the **mean salary is above the median**, because the
right skew exists inside every group too.

## 6.5 Scatter plots — relationships between two columns

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4.5))

pairs = [("experience_years", "annual_salary_inr"),
         ("work_hours_per_week", "satisfaction_score"),
         ("commute_km", "annual_salary_inr")]

for ax, (x, y) in zip(axes, pairs):
    sns.regplot(data=df, x=x, y=y, ax=ax,
                scatter_kws={"alpha": 0.5, "s": 22, "color": "#02707E"},
                line_kws={"color": "red"})
    ax.set_title(f"r = {df[x].corr(df[y]):.3f}", fontsize=11)

plt.suptitle("Strong positive  |  Moderate negative  |  No relationship", fontsize=13, y=1.03)
plt.tight_layout()
plt.show()
```

Read the three graphs left to right:

1. **Rising line** — more experience goes with higher salary.
2. **Falling line** — more working hours goes with **lower** satisfaction. This is a
   **negative** relationship.
3. **Flat line** — commute distance tells us nothing about salary.

The numbers in Part 9 will put an exact value on each of these.

# Part 7: Outlier detection

An **outlier** is a value that lies far away from the rest of the data.

**Finding an outlier is a question, not a verdict.** Before deleting anything, ask *why*
it is there. There are three possibilities:

| Cause | Example | What to do |
|---|---|---|
| Data entry error | Age recorded as 250 | Correct it or remove it |
| Genuine rare value | A CEO's salary | **Keep it** — it is real data |
| A different group mixed in | Executives inside a staff table | Analyse the groups separately |

In our dataset the high salaries are the **second and third** kind. They are real executives,
so deleting them would be wrong — but reporting the mean salary without mentioning them
would also be misleading.

## 7.1 Method 1 — the IQR rule (Tukey's fences)

$$\text{Lower fence} = Q_1 - 1.5 \times IQR \qquad
\text{Upper fence} = Q_3 + 1.5 \times IQR$$

Anything outside the fences is flagged. Because the fences are built from **quartiles**,
the extreme values cannot distort the threshold that is meant to catch them.
This makes the IQR rule **robust**, and it is the safer default for skewed data.

## 7.2 Method 2 — the Z-score rule

$$z = \frac{x - \bar{x}}{s}$$

The z-score says how many standard deviations a value lies from the mean.
The usual cutoff is $|z| > 3$, which follows from the empirical rule: only 0.3% of
normal data lies beyond 3 standard deviations.

**Its weakness:** both $\bar{x}$ and $s$ are themselves inflated by the outliers you are
hunting, so the threshold gets pushed outward and some outliers escape. This effect is
called **masking**.

```python
col = "annual_salary_inr"
values = df[col]

# ---- IQR method ----
Q1, Q3 = values.quantile(0.25), values.quantile(0.75)
IQR = Q3 - Q1
lower_fence = Q1 - 1.5 * IQR
upper_fence = Q3 + 1.5 * IQR

iqr_outliers = df[(values < lower_fence) | (values > upper_fence)]

print(f"COLUMN: {col}")
print("=" * 55)
print(f"Q1                 : {Q1:>14,.2f}")
print(f"Q3                 : {Q3:>14,.2f}")
print(f"IQR                : {IQR:>14,.2f}")
print(f"Lower fence        : {lower_fence:>14,.2f}")
print(f"Upper fence        : {upper_fence:>14,.2f}")
print(f"Outliers found     : {len(iqr_outliers):>14}")
```

```python
# ---- Z-score method on the same column ----
df["salary_z"] = stats.zscore(values)
z_outliers = df[df["salary_z"].abs() > 3]

print(f"Mean               : {values.mean():>14,.2f}")
print(f"Standard deviation : {values.std():>14,.2f}")
print(f"Cutoff |z| > 3     : below {values.mean()-3*values.std():,.0f} "
      f"or above {values.mean()+3*values.std():,.0f}")
print(f"Outliers found     : {len(z_outliers):>14}")
print()
print(f"IQR method flagged     : {len(iqr_outliers)}")
print(f"Z-score method flagged : {len(z_outliers)}")
```

### Why do the two methods disagree?

The Z-score method almost always flags **fewer** points on skewed data. The reason is
**masking**: the executives inflate the standard deviation, which widens the ±3 SD cutoff
until it sits beyond most of the other high salaries. The IQR fences are built from quartiles,
which those same executives cannot move.

**Conclusion:** on skewed data, prefer the IQR rule.

```python
# Look at the actual flagged employees.
# Rebuild the mask from df so the salary_z column added above is included.
iqr_mask = (df[col] < lower_fence) | (df[col] > upper_fence)

out_view = df.loc[iqr_mask, ["emp_id", "department", "education",
                             "experience_years", col, "salary_z"]]              .sort_values(col, ascending=False)

print(f"Employees flagged by the IQR rule: {len(out_view)}")
out_view.head(10).round(2)
```

```python
# Visualise both methods side by side
df["is_outlier"] = (values < lower_fence) | (values > upper_fence)

fig, axes = plt.subplots(1, 2, figsize=(14, 4.8))

sns.scatterplot(data=df, x=df.index, y=col, hue="is_outlier",
                palette={False: "#02707E", True: "#E4572E"}, alpha=0.75, s=30, ax=axes[0])
axes[0].axhline(upper_fence, color="#E4572E", ls="--", lw=1.5)
axes[0].set_title(f"IQR rule — upper fence at {upper_fence:,.0f}")
axes[0].set_xlabel("Employee number")

sns.scatterplot(data=df, x=df.index, y="salary_z",
                hue=df["salary_z"].abs() > 3,
                palette={False: "#00A896", True: "#E4572E"}, alpha=0.75, s=30, ax=axes[1])
axes[1].axhline(3, color="#E4572E", ls="--", lw=1.5)
axes[1].axhline(-3, color="#E4572E", ls="--", lw=1.5)
axes[1].set_title("Z-score rule — cutoff |z| > 3")
axes[1].set_xlabel("Employee number")

plt.tight_layout()
plt.show()
```

```python
# How much do the outliers distort each statistic?
clean_vals = df.loc[~df["is_outlier"], col]

comparison = pd.DataFrame({
    "With outliers": [values.mean(), values.median(), values.std(),
                      values.quantile(.75) - values.quantile(.25),
                      values.skew(), values.kurt()],
    "Without outliers": [clean_vals.mean(), clean_vals.median(), clean_vals.std(),
                         clean_vals.quantile(.75) - clean_vals.quantile(.25),
                         clean_vals.skew(), clean_vals.kurt()],
}, index=["Mean", "Median", "Std Dev", "IQR", "Skewness", "Kurtosis"])

comparison["Change %"] = ((comparison["Without outliers"] - comparison["With outliers"])
                          / comparison["With outliers"].abs() * 100)
comparison.round(2)
```

### The key result of this experiment

Look at the **Change %** column:

- **Mean** and **standard deviation** change a lot. They are **not robust**.
- **Median** and **IQR** barely move. They are **robust**.
- **Skewness** and **kurtosis** collapse once the outliers are removed, which proves that
  those five employees were responsible for almost all of the extreme shape.

This is why any report on salary, house prices or response times should quote the **median**.

```python
# Outlier count for every numeric column
rows = []
for c in num_cols:
    q1, q3 = df[c].quantile(.25), df[c].quantile(.75)
    iqr = q3 - q1
    lo, hi = q1 - 1.5*iqr, q3 + 1.5*iqr
    n_iqr = ((df[c] < lo) | (df[c] > hi)).sum()
    n_z = (np.abs(stats.zscore(df[c])) > 3).sum()
    rows.append({"Column": c, "Lower fence": lo, "Upper fence": hi,
                 "IQR outliers": n_iqr, "Z outliers": n_z,
                 "% flagged": n_iqr / len(df) * 100})

pd.DataFrame(rows).set_index("Column").round(2)
```

# Part 8: Covariance

Covariance measures whether two columns **move together**.

$$\operatorname{cov}(x, y) = \frac{\sum (x_i - \bar{x})(y_i - \bar{y})}{n - 1}$$

### How to understand the formula

For each employee, look at two things at once:

- How far is their experience from the average experience? $(x_i - \bar{x})$
- How far is their salary from the average salary? $(y_i - \bar{y})$

Then multiply those two deviations together:

| Situation | $(x_i-\bar{x})$ | $(y_i-\bar{y})$ | Product |
|---|---|---|---|
| Above average on both | positive | positive | **positive** |
| Below average on both | negative | negative | **positive** |
| Above on one, below on the other | positive | negative | **negative** |

If most employees fall in the first two rows, the products add up to a positive number.
So the **sign** of the covariance tells us the direction of the relationship.

| Sign | Meaning |
|---|---|
| Positive | The two rise together |
| Negative | One rises while the other falls |
| Near zero | No consistent linear relationship |

```python
x = df["experience_years"]
y = df["annual_salary_inr"]

# Step by step, exactly as in the formula
dev_x = x - x.mean()
dev_y = y - y.mean()
products = dev_x * dev_y

print("First 5 employees:")
print(pd.DataFrame({
    "experience": x.head(),
    "dev from mean exp": dev_x.head().round(2),
    "salary": y.head(),
    "dev from mean sal": dev_y.head().round(0),
    "product": products.head().round(0),
}))
print()
print(f"Sum of all products : {products.sum():,.2f}")
print(f"Divide by (n - 1)   : {len(df) - 1}")
print(f"Covariance          : {products.sum() / (len(df) - 1):,.2f}")
print(f"pandas .cov()       : {x.cov(y):,.2f}")
```

```python
# Positive, negative and near-zero covariance in our data
pairs = [("experience_years", "annual_salary_inr"),
         ("age", "experience_years"),
         ("work_hours_per_week", "satisfaction_score"),
         ("commute_km", "satisfaction_score"),
         ("commute_km", "annual_salary_inr")]

print(f"{'Pair':<48}{'Covariance':>18}   Direction")
print("-" * 88)
for a, b in pairs:
    cv = df[a].cov(df[b])
    direction = "POSITIVE - rise together" if cv > 0 else "NEGATIVE - move oppositely"
    print(f"{a + '  vs  ' + b:<48}{cv:>18,.2f}   {direction}")
```

### The big problem with covariance

The covariance between experience and salary is a huge number, and the covariance between
work hours and satisfaction is a small one. It is tempting to conclude that the first
relationship is stronger.

**That conclusion is wrong.** Covariance carries the **units of both columns multiplied
together** — the first is in "years × rupees" and the second in "hours × points".
They are not comparable, and the number changes if we simply change the unit.

```python
# Same relationship, different units
exp_years = df["experience_years"]
exp_months = df["experience_years"] * 12       # identical information
sal = df["annual_salary_inr"]
sal_lakhs = df["annual_salary_inr"] / 100000   # identical information

print("Covariance of experience with salary:")
print(f"  years   vs rupees : {exp_years.cov(sal):>20,.2f}")
print(f"  months  vs rupees : {exp_months.cov(sal):>20,.2f}   (12x bigger)")
print(f"  years   vs lakhs  : {exp_years.cov(sal_lakhs):>20,.2f}   (100000x smaller)")
print()
print("The relationship never changed - only the units did.")
print("So the SIZE of a covariance is meaningless on its own. Only its SIGN is useful.")
```

```python
# The full covariance matrix
df[num_cols].cov().round(1)
```

Notice the diagonal of this matrix: the covariance of a column **with itself** is simply its
**variance**. Covariance is a generalisation of variance to two columns.

The fix for the units problem is to standardise the covariance — which is exactly what
correlation does.

# Part 9: Correlation

Correlation is covariance divided by both standard deviations:

$$r = \frac{\operatorname{cov}(x, y)}{s_x \times s_y}$$

Dividing by $s_x$ cancels the units of $x$, and dividing by $s_y$ cancels the units of $y$.
The result is a pure number with **no units at all**, always between **−1 and +1**.

| Value of $r$ | Meaning |
|---|---|
| +1 | Perfect positive — all points lie exactly on a rising line |
| +0.7 to +1.0 | Very strong positive |
| +0.5 to +0.7 | Strong positive |
| +0.3 to +0.5 | Moderate positive |
| −0.3 to +0.3 | Weak or no linear relationship |
| −0.5 to −0.3 | Moderate negative |
| −1.0 to −0.7 | Very strong negative |
| −1 | Perfect negative — all points lie exactly on a falling line |

The **sign** gives the direction; the **size** gives the strength.

```python
# Proving the formula
x = df["experience_years"]
y = df["annual_salary_inr"]

r_manual = x.cov(y) / (x.std() * y.std())

print(f"Covariance          : {x.cov(y):>18,.2f}")
print(f"Std dev of x        : {x.std():>18,.2f}")
print(f"Std dev of y        : {y.std():>18,.2f}")
print(f"cov / (sx * sy)     : {r_manual:>18.6f}")
print(f"pandas .corr()      : {x.corr(y):>18.6f}")
```

```python
# Correlation is immune to unit changes -- unlike covariance
print("Correlation of experience with salary:")
print(f"  years  vs rupees : {exp_years.corr(sal):.6f}")
print(f"  months vs rupees : {exp_months.corr(sal):.6f}")
print(f"  years  vs lakhs  : {exp_years.corr(sal_lakhs):.6f}")
print()
print("All identical. This is why we report correlation, not covariance.")
```

```python
# The full correlation matrix
corr = df[num_cols].corr()
corr.round(3)
```

```python
plt.figure(figsize=(10, 7.5))
mask = np.triu(np.ones_like(corr, dtype=bool), k=1)
sns.heatmap(corr, mask=mask, annot=True, fmt=".2f", cmap="RdYlGn",
            center=0, vmin=-1, vmax=1, square=True, linewidths=0.5,
            cbar_kws={"label": "Pearson r"})
plt.title("Correlation heatmap", pad=15)
plt.tight_layout()
plt.show()
```

**How to read the heatmap:** dark green means strong positive, dark red means strong negative,
and pale yellow means no relationship. The diagonal is always 1.00 because every column
correlates perfectly with itself.

```python
# Rank every pair by strength
pairs_list = []
for i, a in enumerate(num_cols):
    for b in num_cols[i+1:]:
        r = corr.loc[a, b]
        pairs_list.append({"Variable 1": a, "Variable 2": b, "r": r, "abs_r": abs(r)})

ranked = pd.DataFrame(pairs_list).sort_values("abs_r", ascending=False)
ranked["Strength"] = pd.cut(ranked["abs_r"], [0, 0.3, 0.5, 0.7, 1.0],
                            labels=["Weak", "Moderate", "Strong", "Very strong"])
ranked["Direction"] = np.where(ranked["r"] > 0, "Positive", "Negative")
ranked.drop(columns="abs_r").head(12).round(3).reset_index(drop=True)
```

### What the strongest relationships tell us

Compare these results against what we deliberately built into the data in Part 1:

- **age and experience_years** — the strongest pair, as designed. You cannot have 20 years of
  experience at 25 years of age, so the two are almost locked together.
- **experience_years and annual_salary_inr** — very strong and positive. Experience was the
  dominant term in our salary formula, and the analysis recovered it.
- **work_hours_per_week and satisfaction_score** — a **negative** relationship. We built this
  in with a negative coefficient, and the correlation correctly reports the falling direction.
- **commute_km and annual_salary_inr** — essentially zero. We never connected these two
  columns, and the analysis confirms there is nothing there.

The analysis found what we put in. That is the value of testing your method on synthetic data.

```python
# Pearson vs Spearman
# Pearson uses the values; Spearman uses their RANKS, so it survives outliers
# and detects any rising/falling relationship even if it is not a straight line.
print(f"{'Pair':<52}{'Pearson':>10}{'Spearman':>11}{'Gap':>8}")
print("-" * 82)
for a, b in [("experience_years", "annual_salary_inr"),
             ("age", "annual_salary_inr"),
             ("work_hours_per_week", "satisfaction_score"),
             ("commute_km", "satisfaction_score"),
             ("performance_score", "annual_salary_inr"),
             ("commute_km", "annual_salary_inr")]:
    p = df[a].corr(df[b])
    s = df[a].corr(df[b], method="spearman")
    print(f"{a + '  vs  ' + b:<52}{p:>10.3f}{s:>11.3f}{abs(p-s):>8.3f}")
```

A large gap between Pearson and Spearman is a warning sign. It usually means either
outliers are distorting Pearson, or the relationship is curved rather than straight.

```python
sns.pairplot(df[["experience_years", "age", "performance_score", "annual_salary_inr"]],
             diag_kind="kde", plot_kws={"alpha": 0.4, "s": 18, "color": "#02707E"},
             corner=True)
plt.suptitle("Pair plot — every scatter at once", y=1.01, fontsize=13)
plt.show()
```

## 9.1 Two warnings about correlation

### Warning 1 — correlation only detects **straight lines**

Pearson's $r$ measures **linear** association. A perfect curved relationship can return
$r \approx 0$.

```python
x_curve = np.linspace(-10, 10, 200)
y_curve = x_curve ** 2                 # perfect, exact relationship

sns.scatterplot(x=x_curve, y=y_curve, s=18, color="#02707E")
plt.title(f"y = x²   —   Pearson r = {np.corrcoef(x_curve, y_curve)[0,1]:.4f}")
plt.show()

print("Every point lies exactly on the curve, so y is perfectly predictable from x.")
print("Yet r is almost 0, because the relationship is not a straight line.")
print("LESSON: always draw the scatter plot before trusting r.")
```

### Warning 2 — correlation is **not** causation

A high $r$ means only that two things move together. It does **not** prove that one causes
the other. There are four common explanations:

1. **A confounding variable causes both.**
   Ice cream sales and drowning deaths are strongly correlated. Ice cream does not cause
   drowning — hot weather causes both.
2. **Reverse causality.** The arrow may point the other way.
3. **Coincidence.** Compare enough pairs of variables and some will correlate by pure chance.
4. **Simpson's paradox.** A trend present inside every subgroup can vanish or even reverse
   when the groups are combined.

In our own dataset, `age` and `annual_salary_inr` are correlated — but age does not pay
anyone. **Experience** is the real driver, and experience rises with age. Age is a
confounding variable.

```python
# Demonstrating the confounder
print("age            vs salary :", round(df["age"].corr(df["annual_salary_inr"]), 3))
print("experience     vs salary :", round(df["experience_years"].corr(df["annual_salary_inr"]), 3))
print("age            vs experience:", round(df["age"].corr(df["experience_years"]), 3))
print()
print("Age appears related to salary only because age is related to experience,")
print("and experience is what actually drives salary in this dataset.")
```

# Result and Conclusion

A synthetic dataset of 400 employee records was created and analysed. All required descriptive
statistics were calculated, outliers were detected by two methods, and the relationships
between variables were quantified using covariance and correlation.

### Main conclusions

1. **Choose the average to match the shape.** For symmetric columns such as `age` and
   `work_hours_per_week`, the mean is fine. For skewed columns such as `annual_salary_inr`
   and `commute_km`, report the **median**.

2. **Mode is the only average for text data.** It is the only way to describe `department`,
   `city`, `gender` or `education`.

3. **Variance and standard deviation measure the same thing.** Only the standard deviation
   is in the original units, so only it can be compared against the mean. Use the coefficient
   of variation to compare spread across different units.

4. **Skewness and "mean − median" measure the same asymmetry**, and agreed on every column
   that was genuinely skewed. Where they appeared to disagree (`satisfaction_score`), both
   values were essentially zero — a reminder not to read a direction into a near-zero skew.

5. **High kurtosis warns of outliers before you test for them.** The salary column had by far
   the highest kurtosis, and it was the column with the outliers.

6. **The IQR rule beats the Z-score rule on skewed data.** The Z-score method flagged fewer
   points because the outliers inflate the very standard deviation used to detect them —
   the masking effect.

7. **Median and IQR are robust; mean and standard deviation are not.** Removing 5 of 400
   employees moved the mean noticeably while the median hardly changed.

8. **Covariance gives direction, correlation gives strength.** Covariance changes with the
   unit and so cannot be compared. Correlation is unit-free and bounded by ±1.

9. **A negative correlation is a real finding.** Longer working hours went with lower
   satisfaction, and the sign of `r` captured it correctly.

10. **Correlation is never proof of causation.** Age correlated with salary only because age
    correlates with experience, which is the true driver.
