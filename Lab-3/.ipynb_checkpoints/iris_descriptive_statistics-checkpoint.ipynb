# Descriptive Statistics on the Iris Dataset

**Aim:** To calculate the basic descriptive statistics of a well-known dataset using Python.

**Statistics we will find:**

| # | Statistic | What it tells us |
|---|---|---|
| 1 | Mean | The average value |
| 2 | Median | The middle value |
| 3 | Mode | The most common value |
| 4 | Skewness | Is the data lop-sided? |
| 5 | Kurtosis | Are the tails heavy or light? |
| 6 | Range | Highest − Lowest |
| 7 | Variance | Average squared distance from the mean |
| 8 | Standard deviation | Average distance from the mean |
| 9 | IQR | Spread of the middle 50% |

**Dataset:** Iris — 150 flowers, 4 measurements each, 3 species.
It is the most famous dataset in statistics, collected by Ronald Fisher in 1936.

---
## Step 1: Import libraries and load the data

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (8, 5)

# Iris comes built into seaborn
try:
    df = sns.load_dataset("iris")
except Exception:
    # Backup if there is no internet
    from sklearn.datasets import load_iris
    data = load_iris(as_frame=True)
    df = data.frame
    df.columns = ["sepal_length", "sepal_width", "petal_length", "petal_width", "species"]
    df["species"] = df["species"].map({0: "setosa", 1: "versicolor", 2: "virginica"})

df.head()
```

```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
print()
print(df.dtypes)
```

The four numeric columns are the measurements in centimetres.
`species` is a text column, so we will not calculate statistics on it.

```python
# The columns we will work with
num_cols = ["sepal_length", "sepal_width", "petal_length", "petal_width"]

df[num_cols].head()
```

---
## Step 2: The nine statistics for one column

Let us start with just `sepal_length` so each calculation is clear.

```python
col = df["sepal_length"]

print("1. Mean               :", col.mean())
print("2. Median             :", col.median())
print("3. Mode               :", col.mode()[0])
print("4. Skewness           :", col.skew())
print("5. Kurtosis           :", col.kurt())
print("6. Range              :", col.max() - col.min())
print("7. Variance           :", col.var())
print("8. Standard deviation :", col.std())
print("9. IQR                :", col.quantile(0.75) - col.quantile(0.25))
```

### The same values, rounded and explained

```python
print("Mean   =", round(col.mean(), 2), "cm   -> the average sepal length")
print("Median =", round(col.median(), 2), "cm   -> half the flowers are below this")
print("Mode   =", round(col.mode()[0], 2), "cm   -> the most repeated measurement")
print()
print("Min    =", col.min(), " Max =", col.max())
print("Range  =", round(col.max() - col.min(), 2), "cm   -> Max - Min")
print()
print("Q1     =", col.quantile(0.25))
print("Q3     =", col.quantile(0.75))
print("IQR    =", round(col.quantile(0.75) - col.quantile(0.25), 2), "cm   -> Q3 - Q1")
print()
print("Variance =", round(col.var(), 3))
print("Std Dev  =", round(col.std(), 3), "cm   -> square root of the variance")
```

The mean (5.84) and median (5.80) are very close, so this column is almost balanced.

---
## Step 3: All nine statistics for all four columns

We write one small function and use it on every column.

```python
def describe_column(s):
    """Return the nine descriptive statistics for one column."""
    return pd.Series({
        "Mean": s.mean(),
        "Median": s.median(),
        "Mode": s.mode()[0],
        "Skewness": s.skew(),
        "Kurtosis": s.kurt(),
        "Range": s.max() - s.min(),
        "Variance": s.var(),
        "Std Dev": s.std(),
        "IQR": s.quantile(0.75) - s.quantile(0.25),
    })


stats = df[num_cols].apply(describe_column).round(3)
stats
```

Reading the table: each column of the table is one flower measurement,
and each row is one statistic.

```python
# Turn it around if you prefer one row per measurement
stats.T
```

### Compare with `describe()`

`pandas` has a built-in shortcut that gives some of the same numbers.

```python
df[num_cols].describe().round(3)
```

`describe()` gives count, mean, std, min, Q1, median, Q3 and max —
but **not** mode, skewness, kurtosis, variance, range or IQR.
That is why we wrote our own table above.

---
## Step 4: Understanding skewness and kurtosis

**Skewness** tells us which side the tail is on:

| Value | Meaning |
|---|---|
| Positive | Long tail on the right, mean > median |
| Near 0 | Balanced, mean ≈ median |
| Negative | Long tail on the left, mean < median |

Rule of thumb: below 0.5 is fairly symmetric, 0.5 to 1 is moderate, above 1 is strong.

**Kurtosis** tells us about the tails and the peak.
`pandas` gives *excess* kurtosis, so a perfect normal curve gives **0**.

| Value | Meaning |
|---|---|
| Positive | Sharper peak, heavier tails |
| Near 0 | Similar to a normal curve |
| Negative | Flatter, lighter tails |

```python
shape = pd.DataFrame({
    "Mean": df[num_cols].mean(),
    "Median": df[num_cols].median(),
    "Skewness": df[num_cols].skew(),
    "Kurtosis": df[num_cols].kurt(),
}).round(3)

# Add a plain-English column
shape["Shape"] = np.where(shape["Skewness"] > 0.5, "Right skewed",
                 np.where(shape["Skewness"] < -0.5, "Left skewed", "Almost symmetric"))
shape
```

All four skewness values sit between −0.3 and +0.4, so every column is fairly balanced.
The two **sepal** columns lean slightly to the right, and the two **petal** columns lean
slightly to the left, but none of them cross 0.5.

Notice the kurtosis of `petal_length` (−1.40) and `petal_width` (−1.34).
A strongly negative kurtosis means the distribution is **flat**, not peaked —
a warning sign that we will explain in Step 5.

---
## Step 5: See the statistics in a graph

### Histogram with mean and median marked

```python
sns.histplot(data=df, x="sepal_length", bins=20, kde=True)
plt.axvline(df["sepal_length"].mean(), color="red", linestyle="--", label="Mean")
plt.axvline(df["sepal_length"].median(), color="green", label="Median")
plt.title("Sepal length")
plt.legend()
plt.show()
```

The red and green lines almost overlap. That is what a skewness near zero looks like.

### Histograms of all four columns

```python
fig, axes = plt.subplots(2, 2, figsize=(12, 7))

for ax, c in zip(axes.flatten(), num_cols):
    sns.histplot(data=df, x=c, bins=20, kde=True, ax=ax)
    ax.set_title(f"{c}  (skew = {df[c].skew():.2f})")

plt.tight_layout()
plt.show()
```

`petal_length` and `petal_width` have **two peaks**.
This happens because one species is very different from the other two.
A single mean cannot describe a two-peaked distribution well.

### Box plot — shows median, IQR and outliers

```python
sns.boxplot(data=df[num_cols])
plt.title("Box plot of all four measurements")
plt.ylabel("cm")
plt.show()
```

The box is Q1 to Q3, so the **height of the box is the IQR**.
The line inside is the median. Dots outside the whiskers are outliers.

### Why petal columns have two peaks

```python
sns.boxplot(data=df, x="species", y="petal_length")
plt.title("Petal length by species")
plt.show()
```

Setosa has much smaller petals. That is the reason for the two peaks.

### Statistics for each species separately

```python
df.groupby("species")[num_cols].mean().round(2)
```

```python
# Mean, median and standard deviation of petal_length for each species
df.groupby("species")["petal_length"].agg(["mean", "median", "std", "min", "max"]).round(2)
```

Now the numbers make much more sense. Splitting the data by species gives
three clean groups instead of one confusing mixed distribution.

---
## Step 6: Save the results

```python
stats.T.to_csv("iris_descriptive_statistics.csv")
print("Saved as iris_descriptive_statistics.csv")

stats.T
```

---
## Result

We successfully calculated the mean, median, mode, skewness, kurtosis, range,
variance, standard deviation and IQR for all four numeric columns of the Iris dataset,
and confirmed the results with histograms and box plots.

**Main observations:**

1. All four columns have skewness close to zero, so they are fairly balanced.
2. `petal_length` has the largest variance and the largest range — it varies the most.
3. `sepal_width` has the smallest variance (0.19) and standard deviation (0.44) — it is the most consistent measurement.
4. `petal_length` and `petal_width` have two peaks and strongly negative kurtosis (about −1.4),
   because the setosa species is much smaller than the other two.
5. For `petal_length` the mean (3.76) and median (4.35) differ by more than half a centimetre —
   a sign that one average alone is not enough to describe it.
6. When data contains different groups, calculate the statistics for each group separately.
