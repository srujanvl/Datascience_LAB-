## Step 1: Import libraries and load the data

We use `pandas` for the calculations and `seaborn` for the graphs.

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats

sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (8, 5)

df = pd.read_csv("student_performance.csv")
df.head()
```

```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
print()
print(df.columns.tolist())
```

Each row is one student. The important columns are:

- `branch`, `gender`, `hostel_status` — text columns (categories)
- `study_hours_per_week`, `attendance_pct`, `monthly_expense_inr` — habits
- `internal_marks`, `final_score`, `cgpa` — marks

---
## Step 2: Clean the data

Real data always has problems. This file has three:
missing values, repeated rows, and spelling mistakes in the `gender` column.

```python
# Check for missing values
df.isnull().sum()
```

```python
# Check for repeated rows
print("Duplicate rows:", df.duplicated().sum())
print("Gender labels:", df["gender"].unique())
```

```python
# Fix all three problems
df = df.drop_duplicates()                              # remove repeated rows
df["gender"] = df["gender"].str.strip().str.title()    # fix spelling

df["sleep_hours"] = df["sleep_hours"].fillna(df["sleep_hours"].median())
df["attendance_pct"] = df["attendance_pct"].fillna(df["attendance_pct"].median())

print("Rows now:", len(df))
print("Missing values now:", df.isnull().sum().sum())
print("Gender labels now:", df["gender"].unique())
```

We filled the missing values with the **median** because the median is not affected by very
large or very small values.

---
## Step 3: Mean, median and mode

- **Mean** = average of all values
- **Median** = middle value when sorted
- **Mode** = value that appears most often

```python
print("Mean of final_score  :", df["final_score"].mean())
print("Median of final_score:", df["final_score"].median())
print("Mode of final_score  :", df["final_score"].mode()[0])
```

```python
# All numeric columns at once
num_cols = ["study_hours_per_week", "sleep_hours", "attendance_pct", "commute_km",
            "monthly_expense_inr", "internal_marks", "final_score", "cgpa"]

pd.DataFrame({
    "Mean": df[num_cols].mean(),
    "Median": df[num_cols].median(),
    "Mode": df[num_cols].mode().iloc[0],
}).round(2)
```

### Which one should we use?

Compare the mean and the median:

- If **mean > median**, the data has a few very large values (right skew).
- If **mean < median**, the data has a few very small values (left skew).
- If **mean = median**, the data is balanced.

```python
compare = pd.DataFrame({
    "Mean": df[num_cols].mean(),
    "Median": df[num_cols].median(),
})
compare["Difference"] = compare["Mean"] - compare["Median"]
compare.round(2)
```

```python
# Mode is the only one that works for text columns
print("Most common branch:", df["branch"].mode()[0])
print()
print(df["branch"].value_counts())
```

---
## Step 4: Standard deviation and spread

Two classes can have the same average marks but very different students.
Spread tells us how far the values are from the mean.

- **Range** = highest − lowest
- **Variance** = average of the squared distances from the mean
- **Standard deviation** = square root of variance (same unit as the data)
- **IQR** = Q3 − Q1 (the range of the middle 50% of students)

```python
print("Standard deviation of final_score:", df["final_score"].std().round(2))
print("Variance of final_score          :", df["final_score"].var().round(2))
print("Range of final_score             :", (df["final_score"].max() - df["final_score"].min()).round(2))
```

```python
spread = pd.DataFrame({
    "Min": df[num_cols].min(),
    "Max": df[num_cols].max(),
    "Std Dev": df[num_cols].std(),
    "Variance": df[num_cols].var(),
    "Q1": df[num_cols].quantile(0.25),
    "Q3": df[num_cols].quantile(0.75),
})
spread["IQR"] = spread["Q3"] - spread["Q1"]
spread.round(2)
```

`pandas` can also give most of these in one command:

```python
df[num_cols].describe().round(2)
```

### Skewness

Skewness is a number that tells us if the data leans to one side.

- Positive value = long tail on the right
- Negative value = long tail on the left
- Near zero = balanced

```python
df[num_cols].skew().round(2).sort_values()
```

## Step 6: Finding outliers

An outlier is a value that is very far from the rest. There are two common methods.

**Method 1 — IQR method**

    Lower limit = Q1 - 1.5 * IQR
    Upper limit = Q3 + 1.5 * IQR

Any value outside these limits is an outlier.

**Method 2 — Z-score method**

    z = (value - mean) / standard deviation

If `z` is more than 3 or less than -3, the value is an outlier.

```python
col = "monthly_expense_inr"

Q1 = df[col].quantile(0.25)
Q3 = df[col].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

print("Q1    :", Q1)
print("Q3    :", Q3)
print("IQR   :", IQR)
print("Lower limit:", round(lower, 2))
print("Upper limit:", round(upper, 2))
```

```python
# Which rows are outliers?
outliers = df[(df[col] < lower) | (df[col] > upper)]

print("Number of outliers:", len(outliers))
outliers[["student_id", "branch", col]].sort_values(col, ascending=False).head()
```

```python
# Z-score method on the same column
df["z_score"] = stats.zscore(df[col])

z_outliers = df[df["z_score"].abs() > 3]
print("Outliers by Z-score method:", len(z_outliers))
```

```python
# Mark the outliers and plot them
df["is_outlier"] = (df[col] < lower) | (df[col] > upper)

sns.scatterplot(data=df, x=df.index, y=col, hue="is_outlier",
                palette={False: "teal", True: "red"}, alpha=0.7)
plt.axhline(upper, color="red", linestyle="--")
plt.title("Outliers in monthly expense (red points)")
plt.xlabel("Student number")
plt.show()
```

### Why do outliers matter?

Let us remove them and see what changes.

```python
without = df[~df["is_outlier"]][col]

pd.DataFrame({
    "With outliers": [df[col].mean(), df[col].median(), df[col].std()],
    "Without outliers": [without.mean(), without.median(), without.std()],
}, index=["Mean", "Median", "Std Dev"]).round(2)
```

The **mean** and **standard deviation** change a lot.
The **median** hardly changes.

**Conclusion:** the median is safer to use when the data has outliers.

## Step 7: Covariance and correlation

Both tell us whether two columns move together.

**Covariance** tells us the direction only:

- Positive = both increase together
- Negative = one increases while the other decreases

The problem is that its value changes if we change the unit, so we cannot say if it is big or small.

**Correlation (r)** solves this. It is always between −1 and +1.

| Value of r | Meaning |
|---|---|
| 0.0 to 0.3 | Weak |
| 0.3 to 0.5 | Moderate |
| 0.5 to 0.7 | Strong |
| 0.7 to 1.0 | Very strong |

```python
# Covariance between study hours and final score
print("Covariance:", df["study_hours_per_week"].cov(df["final_score"]).round(3))

# Correlation between the same two columns
print("Correlation:", df["study_hours_per_week"].corr(df["final_score"]).round(3))
```

```python
# Why we use correlation instead of covariance
hours = df["study_hours_per_week"]
minutes = hours * 60          # same data, different unit
score = df["final_score"]

print("Covariance in hours  :", hours.cov(score).round(2))
print("Covariance in minutes:", minutes.cov(score).round(2))
print()
print("Correlation in hours  :", hours.corr(score).round(4))
print("Correlation in minutes:", minutes.corr(score).round(4))
```

The covariance became 60 times bigger just by changing the unit,
but the correlation did not change at all. That is why we use correlation.

```python
# Correlation between all numeric columns
corr = df[num_cols].corr()
corr.round(2)
```

```python
sns.heatmap(corr, annot=True, cmap="RdYlGn", center=0, fmt=".2f")
plt.title("Correlation heatmap")
plt.show()
```

Dark green means strong positive relation. The strongest pair is
`internal_marks` and `final_score`.

### Scatter plots for strong, moderate and weak correlation

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

pairs = [("internal_marks", "final_score"),
         ("study_hours_per_week", "final_score"),
         ("commute_km", "final_score")]

for ax, (x, y) in zip(axes, pairs):
    sns.regplot(data=df, x=x, y=y, ax=ax,
                scatter_kws={"alpha": 0.5, "s": 20},
                line_kws={"color": "red"})
    ax.set_title(f"r = {df[x].corr(df[y]):.2f}")

plt.tight_layout()
plt.show()
```

The first graph shows a clear line, the second shows a rough pattern,
and the third is just a cloud of points with no pattern.

### Important note

**Correlation does not mean causation.**
Two things may increase together without one causing the other.

Example: ice cream sales and drowning cases both increase in summer.
Ice cream does not cause drowning — hot weather causes both.

```python
# Correlation only finds straight-line relations
x = np.linspace(-10, 10, 200)
y = x ** 2

sns.scatterplot(x=x, y=y, s=15)
plt.title(f"Perfect curve, but correlation r = {np.corrcoef(x, y)[0, 1]:.2f}")
plt.show()

print("The relation is perfect, but r is almost 0 because it is not a straight line.")
print("So always draw the graph before trusting the correlation value.")
```

## Step 8: Central Limit Theorem

**Statement:** If we take many random samples from any population and find the mean of each
sample, those means will form a normal (bell-shaped) curve — even if the original data is
not bell-shaped.

**Standard Error** = σ / √n

This means bigger samples give more accurate results.

Let us test this using the `monthly_expense_inr` column, which is not bell-shaped.

```python
# Remove the outliers first so the result is clear
population = df[~df["is_outlier"]][col].values

print("Population size:", len(population))
print("Population mean (mu)   :", population.mean().round(2))
print("Population std (sigma) :", population.std().round(2))
print("Skewness               :", pd.Series(population).skew().round(2), "-> not normal")
```

```python
# Take 2000 samples of each size and find the mean of every sample
np.random.seed(42)

sample_sizes = [2, 5, 30, 100]
means = {}

for n in sample_sizes:
    samples = np.random.choice(population, size=(2000, n))
    means[n] = samples.mean(axis=1)

print("Done. We now have 2000 sample means for each sample size.")
```

```python
# Compare the results
mu = population.mean()
sigma = population.std()

pd.DataFrame({
    "Sample size (n)": sample_sizes,
    "Mean of sample means": [means[n].mean() for n in sample_sizes],
    "Population mean": mu,
    "Standard Error (observed)": [means[n].std() for n in sample_sizes],
    "Standard Error (sigma/sqrt(n))": [sigma / np.sqrt(n) for n in sample_sizes],
    "Skewness": [pd.Series(means[n]).skew() for n in sample_sizes],
}).round(2)
```

```python
# Draw the population and the four sampling distributions
fig, axes = plt.subplots(1, 5, figsize=(18, 3.5))

sns.histplot(population, bins=30, kde=True, ax=axes[0], color="grey")
axes[0].set_title("Population\n(not normal)")

for ax, n in zip(axes[1:], sample_sizes):
    sns.histplot(means[n], bins=30, kde=True, ax=ax)
    ax.set_title(f"n = {n}\nskew = {pd.Series(means[n]).skew():.2f}")
    ax.set_xlabel("Sample mean")

plt.tight_layout()
plt.show()
```

Look at the graphs from left to right:

1. The population is not bell-shaped.
2. As the sample size `n` increases, the curve becomes more and more bell-shaped.
3. By `n = 30` it already looks normal.
4. The curve also becomes **narrower**, because the standard error becomes smaller.

```python
# All four curves together
for n in sample_sizes:
    sns.kdeplot(means[n], label=f"n = {n}")

plt.axvline(mu, color="black", linestyle="--", label="Population mean")
plt.title("Bigger sample size gives a narrower curve")
plt.xlabel("Sample mean")
plt.legend()
plt.show()

print("Standard error at n = 5  :", means[5].std().round(2))
print("Standard error at n = 100:", means[100].std().round(2))
print("It became about", round(means[5].std() / means[100].std(), 1), "times smaller.")
```
