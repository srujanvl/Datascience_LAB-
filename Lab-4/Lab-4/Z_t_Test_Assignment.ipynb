## Quick Reference

| | Z-test | t-test |
|---|---|---|
| Use when | Population σ is **known** | Population σ is **unknown** |
| Formula | $z = \dfrac{\bar{x} - \mu_0}{\sigma / \sqrt{n}}$ | $t = \dfrac{\bar{x} - \mu_0}{s / \sqrt{n}}$, $df = n - 1$ |

| Tail | $H_1$ | p-value (Z) | p-value (t) | Critical value (Z, α = 0.05) |
|---|---|---|---|---|
| Left | $\mu < \mu_0$ | `stats.norm.cdf(z)` | `stats.t.cdf(t, dof)` | −1.645 |
| Right | $\mu > \mu_0$ | `1 - stats.norm.cdf(z)` | `1 - stats.t.cdf(t, dof)` | +1.645 |
| Two | $\mu \neq \mu_0$ | `2 * (1 - stats.norm.cdf(abs(z)))` | `2 * (1 - stats.t.cdf(abs(t), dof))` | ±1.960 |

**Decision rule:** if **p-value ≤ α**, reject $H_0$. Otherwise, fail to reject $H_0$.

For the sample standard deviation use `df['column'].std()`. pandas uses `ddof=1` by default, which is what the t-test needs.

⚠️ The dataset is stored in a variable called `df`. Store the degrees of freedom in a variable called **`dof`**, not `df`, or you will overwrite the dataset.

```python
CURVE = 'navy'
REGION = 'red'
STAT = 'green'

def plot_test(stat, tail, df=None, title=''):
    # df=None means Z-test; otherwise, t-test with the given degrees of freedom
    dist = stats.norm if df is None else stats.t(df)

    lim = max(4.5, abs(stat) + 0.8)
    x = np.linspace(-lim, lim, 600)
    y = dist.pdf(x)

    if tail == 'left':
        crits = [dist.ppf(alpha)]
        region = x <= crits[0]

    elif tail == 'right':
        crits = [dist.ppf(1 - alpha)]
        region = x >= crits[0]

    elif tail == 'two':
        c = dist.ppf(1 - alpha / 2)
        crits = [-c, c]
        region = abs(x) >= c

    else:
        raise ValueError("tail must be 'left', 'right', or 'two'")

    plt.figure(figsize=(9, 5))
    plt.plot(x, y, color=CURVE, lw=2, label='Sampling distribution')

    plt.fill_between(
        x, y,
        where=region,
        color=REGION,
        alpha=0.6,
        label='Rejection region'
    )

    for c in crits:
        plt.axvline(c, color=CURVE, ls=':', lw=1.5)
        plt.text(
            c,
            y.max() * 1.05,
            f'Critical = {c:+.3f}',
            ha='center',
            fontsize=9,
            bbox=dict(
                facecolor='white',
                edgecolor='none',
                pad=1
            ),
            zorder=5
        )

    plt.axvline(
        stat,
        color=STAT,
        lw=2.5,
        label=f'Test statistic = {stat:.3f}'
    )

    plt.ylim(0, y.max() * 1.15)
    plt.title(title)
    plt.xlabel('z' if df is None else f't  (df = {df})')
    plt.ylabel('Density')
    plt.legend(
        loc='upper center',
        bbox_to_anchor=(0.5, -0.15),
        ncol=2,
        frameon=False
    )
    plt.grid(alpha=0.25)
    plt.tight_layout()
    plt.show()


def decide(p_value):
    if p_value <= alpha:
        print(
            f"p-value = {p_value:.4f} ≤ α = {alpha} "
            "→ REJECT H0"
        )
    else:
        print(
            f"p-value = {p_value:.4f} > α = {alpha} "
            "→ FAIL TO REJECT H0"
        )


print("Helper functions ready.")
```

# Part A – Z-Tests (σ known, use all 40 students)

## Q1. Z-Test – Left-Tailed

A health report says college students sleep **7 hours** per night on average, with a known population standard deviation of **σ = 1.2 hours**.
Using `sleep_hours` for **all 40 students**, test at α = 0.05 whether students sleep **less than 7 hours** on average.

**Steps:** (a) write $H_0$, $H_1$ and the tail, (b) find $n$ and $\bar{x}$, (c) compute $z$, (d) find the critical value and the p-value, (e) make a decision and write a conclusion.

*(15 marks)*

```python
# ===================== Q1 – YOUR CODE HERE =====================
# Z-test, left-tailed: column 'sleep_hours', mu_0 = 7, sigma = 1.2, all 40 students
x = df['sleep_hours'].dropna()
mu_0, sigma = 7, 1.2
n, xbar = len(x), x.mean()
z = (xbar - mu_0) / (sigma / np.sqrt(n))
critical_value = stats.norm.ppf(alpha)
p_value = stats.norm.cdf(z)
decision = 'Reject H0' if p_value <= alpha else 'Fail to reject H0'
print(f'n = {n}, x̄ = {xbar:.4f}')
print(f'z = {z:.4f}, critical value = {critical_value:.4f}, p-value = {p_value:.6f}')
print(decision)
```

```python
x = df['sleep_hours']
n = len(x)
xbar = x.mean()

q1_z = (xbar - 7) / (1.2 / np.sqrt(n))
q1_p = stats.norm.cdf(q1_z)

plot_test(
    stat=q1_z,
    tail='left',
    title='Q1: Left-Tailed Z-Test for Sleep Hours'
)

decide(q1_p)
```

**Your answer – Q1**

- **$H_0$:**μ = 7 hours
- **$H_1$:**μ<7 hours
- **Tail (left / right / two):**Left-tailed
- **Test statistic:**z= -1.8578
- **Critical value:**z0.05 = -1.6449
- **p-value:**0.031596
- **Decision (Reject $H_0$ / Fail to reject $H_0$):**Reject H₀
- **Conclusion in words:**At the 5% significance level, there is sufficient evidence that the mean sleep time is less than 7 hours.

## Q4. t-Test – Left-Tailed

*(15 marks)*

```python
# ===================== Q4 – YOUR CODE HERE =====================
# t-test, left-tailed: column 'attendance', Section B only, mu_0 = 83, sigma unknown
x = df.loc[df['section'] == 'B', 'attendance'].dropna()
mu_0 = 83
n, xbar, s = len(x), x.mean(), x.std()
dof = n - 1
t_stat = (xbar - mu_0) / (s / np.sqrt(n))
critical_value = stats.t.ppf(alpha, dof)
p_value = stats.t.cdf(t_stat, dof)
decision = 'Reject H0' if p_value <= alpha else 'Fail to reject H0'
print(f'n = {n}, x̄ = {xbar:.4f}, s = {s:.4f}, dof = {dof}')
print(f't = {t_stat:.4f}, critical value = {critical_value:.4f}, p-value = {p_value:.6f}')
print(decision)
print('SciPy check:', stats.ttest_1samp(x, mu_0, alternative='less'))
```
